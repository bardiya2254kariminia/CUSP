```python
def CUSP_format(df : pd.DataFrame):
    df["Age"] = 0
    for i in range(len(df)):
        df.iloc[i,2] = df.iloc[i,0].split("_")[0]
    df = df.drop("race" ,  axis=1)
    return df

cucp_train_df = CUSP_format(train_df)
cucp_test_df = CUSP_format(test_df)
```

```python
cucp_train_df.to_csv("cucp_train.csv")
cucp_test_df.to_csv("cucp_test.csv")
cucp_train_df.to_numpy()
```

```python
import json
from argparse import  Namespace

class DotDict(dict):
    """A dictionary with dot-access access. Keeps JSON serializability."""
    def __getattr__(self, name):
        val = self.get(name)
        if isinstance(val, dict):
            return DotDict(val)
        elif isinstance(val, list):
            return [DotDict(i) if isinstance(i, dict) else i for i in val]
        return val

    def __setattr__(self, name, value):
        self[name] = value

    def __delattr__(self, name):
        del self[name]


with open("cucp_config.json" , "r") as f:
        config_kwargs = json.load(f)
        args = DotDict(config_kwargs)
args.training_set_kwargs.image_path
```
